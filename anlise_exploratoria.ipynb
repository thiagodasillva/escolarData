```python
numeric_features = dados_df.select_dtypes(include=['number']).columns 

plt.figure(figsize=(20,20))

for i in range(len(numeric_features)):
    plt.subplot(11, 6, i+1)  # Grid 11x6 para 66 gráficos
    sns.kdeplot(x=dados_df[numeric_features[i]], shade=True, color='b')
    plt.xlabel(numeric_features[i])
    plt.tight_layout()  # Ajusta o layout para evitar sobreposição

# Exibindo o gráfico
plt.show()


```
