```python
# ===============================
# 1. VERIFICAR E PREPARAR OS DADOS
# ===============================

# Supondo que seu dataframe se chame 'df'
# Verificar os tipos de dados
print("\n=== Tipos de dados ===")
print(df.dtypes)

# Converter colunas numéricas para garantir que estão no formato correto
colunas_numericas = ['Total', 'Forca_de_Trabalho', 'Ocupados', 'Desocupados', 'Fora_da_Forca',
                    'Taxa_Desocupacao', 'Nivel_Ocupacao', 'Taxa_Participacao']

for col in colunas_numericas:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# Verificar valores nulos
print("\n=== Valores nulos ===")
print(df.isnull().sum())

# Remover linhas com valores nulos (se necessário)
df.dropna(inplace=True)

# ===============================
# 2. ADICIONAR COLUNAS (SE NECESSÁRIO)
# ===============================

# Se precisar recalcular algum indicador (opcional)
df['Taxa_Desocupacao_Recalc'] = df['Desocupados'] / df['Forca_de_Trabalho'] * 100
df['Nivel_Ocupacao_Recalc'] = df['Ocupados'] / df['Total'] * 100
df['Taxa_Participacao_Recalc'] = df['Forca_de_Trabalho'] / df['Total'] * 100

# ===============================
# 3. SALVAR OU EXPORTAR OS DADOS
# ===============================

# Selecionar colunas desejadas (ajuste conforme necessidade)
colunas_finais = [
    'UF', 'Ano', 'Trimestre', 'Trimestre_Ano',
    'Total', 'Forca_de_Trabalho', 'Ocupados', 'Desocupados', 'Fora_da_Forca',
    'Taxa_Desocupacao', 'Nivel_Ocupacao', 'Taxa_Participacao'
]

df_final = df[colunas_finais]

# Salvar em Excel
df_final.to_excel('dados_mercado_trabalho_formatados.xlsx', index=False)

# Salvar em CSV
df_final.to_csv('dados_mercado_trabalho_formatados.csv', index=False, encoding='utf-8-sig')

# ===============================
# 4. VISUALIZAR RESULTADOS
# ===============================

print("\n=== Dados formatados (primeiras linhas) ===")
print(df_final.head())

print("\n=== Resumo estatístico ===")
print(df_final.describe())

# ===============================
# 5. ANÁLISE INICIAL (OPCIONAL)
# ===============================

# Taxa média de desocupação por ano
print("\n=== Taxa média de desocupação por ano ===")
print(df_final.groupby('Ano')['Taxa_Desocupacao'].mean())

# Estados com maior taxa de desocupação média no período
print("\n=== Estados com maior taxa de desocupação média (2012-2021) ===")
print(df_final.groupby('UF')['Taxa_Desocupacao'].mean().sort_values(ascending=False).head(10))
```

```python
# ===============================
# 1. CARREGAR OS DADOS
# ===============================

df = pd.read_csv("indicadores_tabela.csv")

# Renomear a primeira coluna para 'UF'
df.rename(columns={df.columns[0]: "UF"}, inplace=True)

# ===============================
# 2. TRANSFORMAR PARA FORMATO LONGO
# ===============================

# Lista com trimestres
trimestres = []
for i in range(40):  # até 39, pois temos colunas com sufixo até .39
    trimestres.append({
        "Total": f"Total.{i}" if i > 0 else "Total",
        "Forca_de_Trabalho": f"Força de trabalho.{i}" if i > 0 else "Força de trabalho",
        "Ocupados": f"Força de trabalho - ocupada.{i}" if i > 0 else "Força de trabalho - ocupada",
        "Desocupados": f"Força de trabalho - desocupada.{i}" if i > 0 else "Força de trabalho - desocupada",
        "Trimestre": i + 1  # de 1 a 40
    })

# Lista para armazenar os DataFrames de cada trimestre
df_list = []

for t in trimestres:
    temp = df[["UF", t["Total"], t["Forca_de_Trabalho"], t["Ocupados"], t["Desocupados"]]].copy()
    temp.columns = ["UF", "Total", "Forca_de_Trabalho", "Ocupados", "Desocupados"]
    temp["Trimestre_ID"] = t["Trimestre"]
    df_list.append(temp)

# Concatenar todos os trimestres
df_long = pd.concat(df_list, ignore_index=True)

===============================
3. LIMPAR E CONVERTER
===============================

Converter para numérico
for col in ["Total", "Forca_de_Trabalho", "Ocupados", "Desocupados"]:
    df_long[col] = pd.to_numeric(df_long[col], errors="coerce")

# Remover linhas inválidas
df_long.dropna(subset=["Total", "Forca_de_Trabalho", "Ocupados", "Desocupados"], inplace=True)

# Criar colunas de ano e trimestre
df_long["Ano"] = 2012 + (df_long["Trimestre_ID"] - 1) // 4
df_long["Trimestre"] = (df_long["Trimestre_ID"] - 1) % 4 + 1
df_long["Trimestre_Ano"] = "T" + df_long["Trimestre"].astype(str) + "/" + df_long["Ano"].astype(str)

# ===============================
# 4. CALCULAR INDICADORES DA TABELA 4092
# ===============================

df_long["Taxa_Desocupacao"] = df_long["Desocupados"] / df_long["Forca_de_Trabalho"] * 100
df_long["Nivel_Ocupacao"] = df_long["Ocupados"] / df_long["Total"] * 100
df_long["Taxa_Participacao"] = df_long["Forca_de_Trabalho"] / df_long["Total"] * 100
df_long["Fora_da_Forca"] = df_long["Total"] - df_long["Forca_de_Trabalho"]

# ===============================
# 5. VISUALIZAR OU SALVAR
# ===============================

# Selecionar apenas colunas úteis
colunas_finais = [
    "UF", "Ano", "Trimestre", "Trimestre_Ano",
    "Total", "Forca_de_Trabalho", "Ocupados", "Desocupados", "Fora_da_Forca",
    "Taxa_Desocupacao", "Nivel_Ocupacao", "Taxa_Participacao"
]

df_indicadores = df_long[colunas_finais]

# Salvar em Excel
df_indicadores.to_excel("indicadores_tabela_4092.xlsx", index=False)

# Mostrar as 5 primeiras linhas
print("\n=== Indicadores da Tabela 4092 (exemplo) ===")
print(df_indicadores.head())
```
